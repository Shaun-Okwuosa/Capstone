The data is split into `laps`, `race_control`, `results`, `session_status`, `track_status` and `weather`.

Each one of these is then further split by the year from 2021 to 2025.

If they remain split by year, we will potentially have to repeat cleaning tasks for each of the files. 

Therefore, they will be merged.

```python
import numpy as np
import pandas as pd
```

```python
# Loading all of the data into pandas dataframes.

# laps
df_laps_21= pd.read_parquet("combined/laps_2021.parquet")
df_laps_22= pd.read_parquet("combined/laps_2022.parquet")
df_laps_23= pd.read_parquet("combined/laps_2023.parquet")
df_laps_24= pd.read_parquet("combined/laps_2024.parquet")
df_laps_25= pd.read_parquet("combined/laps_2025.parquet")

# race_control
df_rc_21= pd.read_parquet("combined/race_control_2021.parquet")
df_rc_22= pd.read_parquet("combined/race_control_2022.parquet")
df_rc_23= pd.read_parquet("combined/race_control_2023.parquet")
df_rc_24= pd.read_parquet("combined/race_control_2024.parquet")
df_rc_25= pd.read_parquet("combined/race_control_2025.parquet")

# results
df_results_21= pd.read_parquet("combined/results_2021.parquet")
df_results_22= pd.read_parquet("combined/results_2022.parquet")
df_results_23= pd.read_parquet("combined/results_2023.parquet")
df_results_24= pd.read_parquet("combined/results_2024.parquet")
df_results_25= pd.read_parquet("combined/results_2025.parquet")

# session_status
df_ss_21= pd.read_parquet("combined/session_status_2021.parquet")
df_ss_22= pd.read_parquet("combined/session_status_2022.parquet")
df_ss_23= pd.read_parquet("combined/session_status_2023.parquet")
df_ss_24= pd.read_parquet("combined/session_status_2024.parquet")
df_ss_25= pd.read_parquet("combined/session_status_2025.parquet")

# track_status
df_ts_21= pd.read_parquet("combined/track_status_2021.parquet")
df_ts_22= pd.read_parquet("combined/track_status_2022.parquet")
df_ts_23= pd.read_parquet("combined/track_status_2023.parquet")
df_ts_24= pd.read_parquet("combined/track_status_2024.parquet")
df_ts_25= pd.read_parquet("combined/track_status_2025.parquet")

# weather
df_weather_21= pd.read_parquet("combined/weather_2021.parquet")
df_weather_22= pd.read_parquet("combined/weather_2022.parquet")
df_weather_23= pd.read_parquet("combined/weather_2023.parquet")
df_weather_24= pd.read_parquet("combined/weather_2024.parquet")
df_weather_25= pd.read_parquet("combined/weather_2025.parquet")
```

The first step should be ensuring that the column names remain the same across all 5 years.

```python
names= [df_laps_21, df_laps_22, df_laps_23, df_laps_24, df_laps_25] # All the dataframes for laps.

for i in range(len(names)):
    for n in range(len(names)):
        if n != i:

            # Making a direct comparison of the entire list because the column orders need to match so that they can be concatenated.
            if names[i].columns.to_list() != names[n].columns.to_list(): 
                print(f"{names[i]} and {names[n]} have different columns.")
```

All columns in `laps` across all 5 years have the same names.

```python
# Concatenating the dataframes.
df_laps= pd.concat([df_laps_21, df_laps_22, df_laps_23, df_laps_24, df_laps_25], ignore_index= True)
df_race_control= pd.concat([df_rc_21, df_rc_22, df_rc_23, df_rc_24, df_rc_25], ignore_index= True)
df_results= pd.concat([df_results_21, df_results_22, df_results_23, df_results_24, df_results_25], ignore_index= True)
df_session_status= pd.concat([df_ss_21, df_ss_22, df_ss_23, df_ss_24, df_ss_25], ignore_index= True)
df_track_session= pd.concat([df_ts_21, df_ts_22, df_ts_23, df_ts_24, df_ts_25], ignore_index= True)
df_weather= pd.concat([df_weather_21, df_weather_22, df_weather_23, df_weather_24, df_weather_25], ignore_index= True)
```

All of them have been merged.

Thes dataframes will be saved as `.parquet` files because `.csv` would be really slow to work on.

```python
df_laps.to_parquet("Data/all_laps.parquet")
df_race_control.to_parquet("Data/all_race_control.parquet")
df_results.to_parquet("Data/all_results.parquet")
df_session_status.to_parquet("Data/all_session_status.parquet")
df_track_session.to_parquet("Data/all_track_status.parquet")
df_weather.to_parquet("Data/all_weather.parquet")
```
